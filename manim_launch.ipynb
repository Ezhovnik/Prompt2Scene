```python
%%manim -qm --verbosity WARNING GreenSquareToOrangeTriangle

#manim-код вставляйте ниже. После WARNING в команде выше напишите название класса с методом construct

from manim import *

class GreenSquareToOrangeTriangle(Scene):
    def construct(self):
        self.camera.background_color = BLACK

        square = Square(color=GREEN).scale(3)
        triangle = Triangle(color=ORANGE).scale(2)

        self.play(Create(square))
        self.wait()
        self.play(Transform(square, triangle))
        self.wait()
```
